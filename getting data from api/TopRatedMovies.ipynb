```python
import pandas as pd 
import requests
import os
import time
from dotenv import load_dotenv
```

```python
load_dotenv()

API_KEY = os.getenv("API_KEY")

response = requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key={API_KEY}&language=en-US&page=1')
```

```python
df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df = pd.DataFrame()
```

```python
all_movies = [] 

for i in range(2, 429):
    url = f'https://api.themoviedb.org/3/movie/top_rated?api_key={API_KEY}&language=en-US&page={i}'
    
    try:
        response = requests.get(url, timeout=10)
        
        if response.status_code == 200:
            data = response.json()
            page_movies = data.get('results', [])
            if page_movies:
                all_movies.extend(page_movies)
                print(f"Successfully fetched page {i}")
            else:
                print(f"No results on page {i}. Stopping.")
                break
        else:
            print(f"Failed to fetch page {i}. Status code: {response.status_code}")
           
    except requests.exceptions.RequestException as e:
        print(f"An error occurred on page {i}: {e}")
    
    time.sleep(0.1) 

print("\nFinished fetching data. Creating DataFrame...")

df = pd.DataFrame(all_movies)
df.head()
```

```python
df = df[['id','title','overview','release_date','popularity','vote_average','vote_count']]
df.head()
```
